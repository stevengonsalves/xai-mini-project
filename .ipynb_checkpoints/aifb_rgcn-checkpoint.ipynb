## 1. Import Libraries

This section imports all libraries required for loading RDF data, processing graphs, building the R-GCN model, training the model, and generating explanations.

Random seeds are also set to make the results reproducible.

````python
# Count occurrences of items
from collections import Counter

# Read compressed files
import gzip

# Numerical operations
import numpy as np

# Operating system utilities
import os

# Random number generation
import random

# PyTorch library
import torch

# Neural network functions such as ReLU and CrossEntropyLoss
import torch.nn.functional as F

# Data analysis library
import pandas as pd

# RDF graph processing library
import rdflib as rdf

# Graph transformations from PyTorch Geometric
import torch_geometric.transforms as T

# Graph data structure
from torch_geometric.data import Data

# Fast Relational Graph Convolution layer
from torch_geometric.nn import FastRGCNConv

# Utility function for sorting indices
from torch_geometric.utils import index_sort

# GNN explanation tools
from torch_geometric.explain import Explainer, GNNExplainer

# Configuration for explanation settings
from torch_geometric.explain.config import ModelConfig

# Set random seeds for reproducibility
SEED = 42
os.environ["PYTHONHASHSEED"] = str(SEED)
os.environ["CUBLAS_WORKSPACE_CONFIG"] = ":4096:8"

def seed_everything(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.benchmark = False
    torch.backends.cudnn.deterministic = True
    torch.use_deterministic_algorithms(True, warn_only=True)

seed_everything(SEED)

print("Libraries loaded successfully")
````

## 2. Load RDF Dataset

The AIFB RDF knowledge graph is loaded using RDFLib.
The graph contains RDF triples in the form:
Subject → Predicate → Object
After loading, the total number of triples is displayed.

````python
# Path handling library
from pathlib import Path

# RDF processing library
import rdflib as rdf

# Location of the RDF graph file
graph_file = Path("data") / "Entities" / "aifb" / "raw" / "aifbfixed_complete.n3"

# Create RDF graph object
g = rdf.Graph()

# Load RDF graph from file
g.parse(location=str(graph_file), format="n3")

# Print total number of triples
print("Number of triples:", len(g))
````

### Dataset Statistics

The RDF graph contains 29,226 triples.

Each triple represents a relationship between two entities in the knowledge graph.

Examples include:

- Person → member → Research Group
- Person → publication → Publication
- Project → carriedOutBy → Research Group

These relationships form the graph structure used by the R-GCN model.

### Dataset Statistics
Basic statistics are computed to understand the size and structure of the RDF knowledge graph.

````python
# Get all unique subjects from the RDF graph
subjects = set(g.subjects())

# Get all unique objects from the RDF graph
objects = set(g.objects())

# Get all unique predicates (relations) from the RDF graph
predicates = set(g.predicates())

# Print the number of unique subjects
print("Number of subjects:", len(subjects))

# Print the number of unique objects
print("Number of objects:", len(objects))

# Print the number of unique predicates
print("Number of predicates:", len(predicates))
````

### Interpretation

The RDF graph contains:

- 2,829 unique subjects
- 8,096 unique objects
- 47 unique predicates (relationship types)

This shows that the dataset contains thousands of entities connected through 47 different types of relationships.

These entities and relationships form the knowledge graph that will be used to build the R-GCN model.

## 3. RDF to Graph Conversion
RDF triples are converted into graph nodes, edges, and relation types suitable for PyTorch Geometric.

````python
from collections import defaultdict

# Create a deterministically ordered list containing all unique entities
nodes = sorted(set(subjects).union(set(objects)), key=str)

# Assign a stable unique ID to each node
node_to_idx = {node: idx for idx, node in enumerate(nodes)}

# Assign a stable unique ID to each relation type
predicates = sorted(predicates, key=str)
relation_to_idx = {rel: idx for idx, rel in enumerate(predicates)}

# Lists to store graph edges and relation types
edge_index = []
edge_type = []

# Convert RDF triples into graph edges in a deterministic order
for s, p, o in sorted(g, key=lambda triple: tuple(map(str, triple))):

    # Store source node and target node
    edge_index.append([
        node_to_idx[s],
        node_to_idx[o]
    ])

    # Store relation type
    edge_type.append(
        relation_to_idx[p]
    )

# Convert lists into PyTorch tensors
edge_index = torch.tensor(edge_index).t().contiguous()
edge_type = torch.tensor(edge_type)

# Display graph information
print("Edge index shape:", edge_index.shape)
print("Edge type shape:", edge_type.shape)
print("Number of nodes:", len(nodes))
print("Number of relations:", len(predicates))
````

### Graph Conversion Results

The RDF knowledge graph was successfully converted into a graph representation that can be processed by PyTorch Geometric.

Results:

- Edge Index Shape: (2, 29,226)
- Edge Type Shape: (29,226)
- Number of Nodes: 8,285
- Number of Relation Types: 47

The edge index stores the source and target node for every connection in the graph.

The edge type tensor stores the relation type associated with each edge.

A total of 8,285 unique entities were converted into graph nodes, and 47 unique predicates were converted into relation types.

This graph structure will be used as the input for the R-GCN model.

## 4. Label Preparation
The classification labels are loaded from the AIFB dataset and mapped to graph nodes for node classification.

````python
task_file = Path("data") / "Entities" / "aifb" / "raw" / "completeDataset.tsv"

df = pd.read_csv(task_file, sep="\t")

print("Number of labeled instances:", len(df))
print("Columns:", list(df.columns))
print("Number of classes:", df["label_affiliation"].nunique())
````

### Label Preparation Results

The classification labels were successfully loaded from the AIFB dataset.

Results:

- Number of labeled instances: 176
- Number of classes: 4
- Columns available:
  - id
  - person
  - label_affiliation

Each row represents a person entity together with its research group affiliation.

The `label_affiliation` column contains the target class that the R-GCN model will learn to predict.

Since there are 4 unique affiliation labels, this is a 4-class node classification problem.

Only 176 person entities have known labels and can therefore be used for supervised learning and evaluation.

## 5. Label Encoding

The affiliation labels are converted into numerical class indices. A target tensor is then created, where each labeled person node is assigned its corresponding class label.

````python
# Create deterministic label mapping
labels = sorted(df['label_affiliation'].unique())
label_to_idx = {label: idx for idx, label in enumerate(labels)}

# Create target tensor
y = torch.full((len(nodes),), -1, dtype=torch.long)

for _, row in df.iterrows():
    person = rdf.URIRef(row['person'])
    label = row['label_affiliation']
    
    if person in node_to_idx:
        y[node_to_idx[person]] = label_to_idx[label]

print("Number of classes:", len(labels))
print("Labeled nodes:", (y != -1).sum().item())
````

### Label Encoding Results

The affiliation labels were converted into numerical class indices so that they can be used by the neural network.

Results:

- Number of classes: 4
- Number of labeled nodes: 176

A target tensor `y` was created for all graph nodes.

For each labeled person node:
- The affiliation label was converted into a numerical class ID.
- The corresponding node received this class ID in the target tensor.

Nodes without labels were assigned a value of `-1`.

This encoded target tensor will be used during training to teach the R-GCN model how to classify person entities into their correct research affiliations.

## 6. Node Feature Construction

Degree-based node features are generated using OneHotDegree. This provides structural information about each node while keeping the feature dimensionality manageable.

````python
# Create degree-based node features using one-hot encoding
transform = T.OneHotDegree(max_degree=1999, cat=False)

# Create a temporary graph object with edges and number of nodes
tmp_data = Data(edge_index=edge_index, num_nodes=len(nodes))

# Apply OneHotDegree transformation
tmp_data = transform(tmp_data)

# Extract the generated feature matrix
x = tmp_data.x

# Identify feature columns that contain at least one non-zero value
mask = x.abs().sum(dim=0) > 0

# Remove feature columns that are entirely zero
x = x[:, mask]

# Display the final feature matrix size
print("Feature matrix shape:", x.shape)
````

````python
# Check labels
print("Total nodes:", len(nodes))
print("Feature matrix:", x.shape)

print("Labeled nodes:", (y != -1).sum().item())
print("Unlabeled nodes:", (y == -1).sum().item())

# See label values and counts
unique_labels, counts = torch.unique(y, return_counts=True)

print("\nLabel distribution:")
for label, count in zip(unique_labels, counts):
    print(f"Label {label.item()}: {count.item()} nodes")
````

### Node Feature Construction Results

Node features were generated using the OneHotDegree transformation.

Results:

- Number of nodes: 8,285
- Number of node features: 72
- Feature matrix shape: (8285, 72)

Each node is represented by a 72-dimensional feature vector based on its graph degree.

The degree of a node indicates how many connections it has to other nodes in the RDF graph.

OneHotDegree converts each degree value into a one-hot encoded vector, allowing the R-GCN model to use structural information about nodes during training.

Columns that contained only zeros were removed to reduce the feature dimensionality and improve efficiency.

The resulting feature matrix is used as the input feature representation for the R-GCN model.

## 7. Train-Test Split

The labeled nodes are divided into training and testing sets using an 80/20 split. This allows the model to be trained on one subset and evaluated on unseen nodes.

````python
# Import function for splitting data into training and testing sets
from sklearn.model_selection import train_test_split

# Get indices of all labeled nodes
# Nodes with label -1 are unlabeled and ignored
labeled_idx = (y != -1).nonzero(as_tuple=True)[0]

# Split labeled nodes into 80% training and 20% testing
# random_state ensures reproducible results
# stratify keeps class distribution balanced
train_idx, test_idx = train_test_split(
    labeled_idx.numpy(),
    test_size=0.2,
    random_state=SEED,
    stratify=y[labeled_idx].numpy()
)

# Create boolean masks for all graph nodes
train_mask = torch.zeros(len(nodes), dtype=torch.bool)
test_mask = torch.zeros(len(nodes), dtype=torch.bool)

# Mark training nodes
train_mask[train_idx] = True

# Mark testing nodes
test_mask[test_idx] = True

# Display number of training and testing samples
print("Train samples:", train_mask.sum().item())
print("Test samples:", test_mask.sum().item())

# Verify that only labeled nodes are used in the split masks
assert torch.all(y[train_mask] != -1)
assert torch.all(y[test_mask] != -1)
assert not torch.any(train_mask & test_mask)
assert train_mask.sum().item() + test_mask.sum().item() == labeled_idx.numel()
````

### Train-Test Split Results

The dataset contains 176 labeled person nodes.

An 80/20 train-test split was applied:

- Training samples: 140
- Test samples: 36

The training set is used to learn the classification patterns, while the test set is used to evaluate the model on unseen nodes.

Stratified sampling was used to preserve the distribution of the four affiliation classes in both subsets.

Boolean masks (`train_mask` and `test_mask`) were created to indicate which nodes belong to the training and testing sets.

## 8. Creating the Graph Data Object

````python
# Create the final PyTorch Geometric data object
# It contains node features, graph edges, relation types, and labels
data = Data(
    x=x,
    edge_index=edge_index,
    edge_type=edge_type,
    y=y
)

# Add training mask to identify training nodes
data.train_mask = train_mask

# Add testing mask to identify testing nodes
data.test_mask = test_mask

# Display graph information
print(data)
````

### Graph Data Object Results

The final graph was stored in a PyTorch Geometric `Data` object.

Output:

Data(x=[8285, 72], edge_index=[2, 29226], y=[8285], edge_type=[29226], train_mask=[8285], test_mask=[8285])

- The graph contains 29,226 edges connecting the nodes.

This `Data` object contains all information required by the R-GCN model for node classification.

## 9. R-GCN Model

A two-layer Relational Graph Convolutional Network is used to classify person nodes into research affiliation classes.

````python
# Define the Relational Graph Convolutional Network
class RGCN(torch.nn.Module):

    # Initialize model layers
    def __init__(self, in_channels, hidden_channels, out_channels, num_relations):
        super().__init__()

        # First R-GCN layer
        self.conv1 = FastRGCNConv(
            in_channels,
            hidden_channels,
            num_relations=num_relations
        )

        # Second R-GCN layer
        self.conv2 = FastRGCNConv(
            hidden_channels,
            out_channels,
            num_relations=num_relations
        )

    # Forward pass through the network
    def forward(self, x, edge_index, edge_type):

        # First graph convolution layer
        x = self.conv1(x, edge_index, edge_type)

        # Apply ReLU activation
        x = F.relu(x)

        # Second graph convolution layer
        x = self.conv2(x, edge_index, edge_type)

        return x


# Reset seeds immediately before model initialization
seed_everything(SEED)

# Create R-GCN model
model = RGCN(
    in_channels=data.num_node_features,   # 72 node features
    hidden_channels=16,                   # hidden layer size
    out_channels=len(labels),             # 4 output classes
    num_relations=len(predicates)         # 47 relation types
)

# Adam optimizer for training
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.01
)

# Display model architecture
print(model)
````

## R-GCN Model Interpretation

A two-layer Relational Graph Convolutional Network (R-GCN) is created for node classification.

The model uses graph structure, node features, and relation types to learn representations of nodes in the RDF knowledge graph.

Architecture:

- Input features: 72
- Hidden layer size: 16
- Output classes: 4
- Relation types: 47

The first R-GCN layer transforms the input node features into hidden representations.

A ReLU activation function introduces non-linearity.

The second R-GCN layer produces class scores for the four affiliation classes.

The Adam optimizer with a learning rate of 0.01 is used for training.

## 10. Training and Evaluation Functions

The training function updates the R-GCN model parameters using cross-entropy loss and the Adam optimizer.

The evaluation function measures classification accuracy on both the training and test sets to assess model performance.

````python
# Training function
def train():

    # Set model to training mode
    model.train()

    # Clear previous gradients
    optimizer.zero_grad()

    # Forward pass through the R-GCN
    out = model(data.x, data.edge_index, data.edge_type)

    # Compute loss using only training nodes
    loss = F.cross_entropy(
        out[data.train_mask],
        data.y[data.train_mask]
    )

    # Backpropagation
    loss.backward()

    # Update model parameters
    optimizer.step()

    # Return loss value
    return loss.item()


# Disable gradient calculation during evaluation
@torch.no_grad()
def test():

    # Set model to evaluation mode
    model.eval()

    # Generate predictions
    out = model(data.x, data.edge_index, data.edge_type)

    # Select class with highest score
    pred = out.argmax(dim=1)

    # Count correct training predictions
    train_correct = (
        pred[data.train_mask] == data.y[data.train_mask]
    ).sum()

    # Calculate training accuracy
    train_acc = int(train_correct) / int(data.train_mask.sum())

    # Count correct test predictions
    test_correct = (
        pred[data.test_mask] == data.y[data.test_mask]
    ).sum()

    # Calculate test accuracy
    test_acc = int(test_correct) / int(data.test_mask.sum())

    # Return both accuracies
    return train_acc, test_acc
    
````

## Training and Evaluation Functions

The training function updates the R-GCN model parameters using cross-entropy loss and the Adam optimizer.

The evaluation function measures classification accuracy on both the training and test sets to assess model performance.

During training:

- The model processes node features and graph relations.
- Cross-entropy loss is computed on training nodes only.
- Gradients are calculated using backpropagation.
- The optimizer updates the model weights.

During evaluation:

- Predictions are generated for all nodes.
- Training accuracy and test accuracy are calculated separately.
- Accuracy is computed as the percentage of correctly classified nodes.

## 11. Model Training

The R-GCN model is trained for 100 epochs. During training, the loss, training accuracy, and test accuracy are printed every 10 epochs.

````python
# Train the model for 100 epochs
for epoch in range(1, 101):

    # Perform one training step and get the loss
    loss = train()

    # Evaluate the model every 10 epochs
    if epoch % 10 == 0:

        # Calculate training and test accuracy
        train_acc, test_acc = test()

        # Print training progress
        print(
            f"Epoch: {epoch:03d}, "
            f"Loss: {loss:.4f}, "
            f"Train Acc: {train_acc:.4f}, "
            f"Test Acc: {test_acc:.4f}"
        )
````

## 12. Training Results

The R-GCN model achieved strong performance on the AIFB dataset.

- Final Training Accuracy: **97.86%**
- Final Test Accuracy: **94.44%**

The training loss decreased from **0.5551** at epoch 10 to **0.0696** at epoch 100, showing that the model learned meaningful relational patterns from the RDF graph.

The test accuracy fluctuated slightly during training, reaching **97.22%** at epochs 30 and 40 before stabilizing at **94.44%** by epoch 100. This is expected because the AIFB test set contains only **36 labeled nodes**, so one changed prediction affects the test accuracy by about **2.78 percentage points**.

The final result is reproducible after fixing the random seeds, deterministic train-test split, stable RDF node ordering, stable relation ordering, and deterministic model initialization.

## 13. Selecting a Node for Explanation

To explain the model prediction, a correctly classified test node is selected.

Among the correctly classified test nodes, the node with the highest confidence score is chosen for GNNExplainer.

````python
# Set model to evaluation mode
model.eval()

# Get predictions for all nodes
with torch.no_grad():
    out = model(data.x, data.edge_index, data.edge_type)

# Predicted class for each node
pred = out.argmax(dim=1)

# Convert model outputs into probabilities
prob = torch.softmax(out, dim=1)

# Select test nodes where prediction matches the true label
correct_test_nodes = test_idx[
    pred[test_idx] == data.y[test_idx]
]

# Print number of correctly classified test nodes
print("Correct test nodes:", len(correct_test_nodes))

# Print first 10 correctly classified node indices
print(correct_test_nodes[:10])
````

## 14. Choosing the Most Confident Prediction

The confidence scores of all correctly classified test nodes are calculated.

The node with the highest confidence score is selected for explanation because the model is most certain about this prediction.

````python
# Get the highest probability (confidence) for each correctly classified node
confidences = prob[correct_test_nodes].max(dim=1).values

# Find the position of the node with the highest confidence
best_pos = confidences.argmax()

# Get the actual graph node index
node_idx = int(correct_test_nodes[best_pos])

# Display information about the selected node
print("Selected node:", node_idx)

# Print the predicted class
print("Prediction:", pred[node_idx].item())

# Print the true class label
print("True label:", data.y[node_idx].item())

# Print the prediction confidence score
print("Confidence:", confidences[best_pos].item())
````

### Interpretation

The selected node is node **5710**, which was chosen because it has the highest prediction confidence among the correctly classified test nodes.

The model predicted class **2**, and the true label is also **2**, meaning the prediction is correct.

The confidence score is **1.0**, which indicates that the model is extremely confident in this prediction.

Because this node is both correctly classified and predicted with maximum confidence, it is a suitable candidate for explanation using GNNExplainer. The explanation helps identify which RDF relationships and graph connections contributed most strongly to the model's decision..

## 15. GNNExplainer Configuration

GNNExplainer is used to identify the most important graph relationships responsible for the prediction of the selected node.

The explainer learns an edge importance mask and highlights the connections that contribute most to the final classification.

````python
# Reset seeds immediately before GNNExplainer mask initialization
seed_everything(SEED)

# Create a GNNExplainer object

explainer = Explainer(

    # Trained R-GCN model to be explained
    model=model,

    # Use GNNExplainer algorithm with 200 optimization epochs
    algorithm=GNNExplainer(epochs=200),

    # Explain the model's prediction
    explanation_type='model',

    # Do not generate node feature importance scores
    node_mask_type=None,

    # Generate importance scores for graph edges
    edge_mask_type='object',

    # Configuration of the prediction task
    model_config=ModelConfig(

        # Multi-class classification problem
        mode='multiclass_classification',

        # Node classification task
        task_level='node',

        # Model returns raw logits
        return_type='raw'
    ),
)
````

## 16. Generating the Explanation

The explanation is generated for the selected node using GNNExplainer. The explainer produces edge importance scores that show which graph relationships influenced the prediction most.

````python
# Reset seeds immediately before generating the explanation
seed_everything(SEED)

# Generate explanation for the selected node

explanation = explainer(

    # Node feature matrix
    data.x,

    # Graph connections
    data.edge_index,

    # Relation type for each edge
    edge_type=data.edge_type,

    # Node to be explained
    index=node_idx
)

# Display explanation object
print(explanation)
````

### Interpretation

The explanation object was successfully generated by GNNExplainer.

The output indicates that:

- `edge_mask=[29226]` means GNNExplainer calculated an importance score for each of the 29,226 edges in the graph.
- `prediction=[8285, 4]` contains the model prediction scores for all 8,285 nodes across 4 classes.
- `target=[8285]` stores the target labels associated with the nodes.
- `x=[8285, 72]` confirms that the node feature matrix contains 8,285 nodes and 72 features per node.
- `edge_index=[2, 29226]` represents the graph structure with 29,226 edges.
- `edge_type=[29226]` stores the relation type for each edge.

Most importantly, the explanation contains the learned `edge_mask`, which assigns an importance score to every edge in the graph. These scores will be used to identify the relationships that contributed most strongly to the prediction of the selected node.

## 17. Extracting the Most Important Edges

GNNExplainer assigns an importance score to every edge in the graph.

The edges with the highest scores are considered the most influential relationships for the prediction of the selected node. Therefore, the top 10 highest-scoring edges are extracted for further analysis.

````python
# Get edge importance scores produced by GNNExplainer
edge_scores = explanation.edge_mask

# Select the 10 most important edges
top_edges = edge_scores.topk(10)

# Display the highest importance scores
print("Top edge scores:")
print(top_edges.values)

print()

# Display the corresponding edge indices
print("Top edge indices:")
print(top_edges.indices)
````

## 18. Human-Readable Explanation

The top-ranked edges are converted back into RDF triples.

This allows us to identify the specific entities and relationships that contributed most strongly to the model prediction and makes the explanation easier to interpret.

````python
# Create reverse mappings from IDs back to original entities and relations
idx_to_node = {idx: node for node, idx in node_to_idx.items()}
idx_to_relation = {idx: rel for rel, idx in relation_to_idx.items()}

# Convert top edge indices and scores into Python lists
top_edge_indices = top_edges.indices.tolist()
top_edge_values = top_edges.values.tolist()

# Loop through the 10 most important edges
for rank, (edge_id, score) in enumerate(
        zip(top_edge_indices, top_edge_values), start=1):

    # Get source node, target node, and relation IDs
    src_idx = int(edge_index[0, edge_id])
    dst_idx = int(edge_index[1, edge_id])
    rel_idx = int(edge_type[edge_id])

    # Convert IDs back to readable RDF entities and relations
    src = idx_to_node[src_idx]
    rel = idx_to_relation[rel_idx]
    dst = idx_to_node[dst_idx]

    # Display the explanation in RDF triple format
    print(f"\nRank {rank}")
    print(f"Importance score: {score:.4f}")
    print(f"Subject: {src}")
    print(f"Predicate: {rel}")
    print(f"Object: {dst}")
````

Since these relationships naturally describe academic collaborations and institutional associations within the AIFB knowledge graph, the explanation is both meaningful and interpretable.

## Evaluation of the Explanation

````python
# Quantitative evaluation of the GNNExplainer explanation

model.eval()

# Number of selected explanation edges
num_explanation_edges = top_edges.indices.numel()

# Total number of edges in the graph
num_total_edges = data.edge_index.size(1)

# Sparsity
sparsity = 1 - (num_explanation_edges / num_total_edges)

# Original prediction probability for the selected node
with torch.no_grad():
    original_out = model(data.x, data.edge_index, data.edge_type)
    original_prob = torch.softmax(original_out, dim=1)[node_idx, pred[node_idx]].item()

# Remove the top explanation edges from the graph
remove_mask = torch.ones(num_total_edges, dtype=torch.bool)
remove_mask[top_edges.indices] = False

edge_index_removed = data.edge_index[:, remove_mask]
edge_type_removed = data.edge_type[remove_mask]

# Prediction after removing explanation edges
with torch.no_grad():
    removed_out = model(data.x, edge_index_removed, edge_type_removed)
    removed_prob = torch.softmax(removed_out, dim=1)[node_idx, pred[node_idx]].item()

# Fidelity-
fidelity_minus = original_prob - removed_prob

# Create evaluation table
explanation_eval = pd.DataFrame({
    "Metric": [
        "Explained node",
        "Original prediction class",
        "Original prediction probability",
        "Probability after removing explanation edges",
        "Fidelity-",
        "Total graph edges",
        "Explanation edges",
        "Sparsity"
    ],
    "Value": [
        node_idx,
        pred[node_idx].item(),
        round(original_prob, 4),
        round(removed_prob, 4),
        round(fidelity_minus, 4),
        num_total_edges,
        num_explanation_edges,
        f"{sparsity * 100:.2f}%"
    ]
})

explanation_eval
````

### Evaluation of the Explanation

The explanation was evaluated using two automatic metrics: **sparsity** and **fidelity-**.

Sparsity measures how compact the explanation is. GNNExplainer selected only the top **10** edges out of **29,226** graph edges, meaning the explanation uses only a very small part of the full graph.

Fidelity- measures how important the explanation edges are for the prediction. To compute this, the top explanation edges were removed from the graph and the model prediction was computed again. If the prediction probability drops after removing these edges, it means that the explanation edges were important for the model's decision.

A high sparsity value means the explanation is concise. A positive fidelity- value means the selected explanation edges contributed to the original prediction.

Together, these metrics show whether the explanation is both compact and relevant to the model prediction.

The RDF nodes, relations, and triples were sorted before assigning numerical IDs. This prevents node indices from changing after a kernel restart.
